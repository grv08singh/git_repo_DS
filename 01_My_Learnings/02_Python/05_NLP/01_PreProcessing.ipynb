## remove URL

```python
def remove_url(text):
    pattern = re.compile(r'http\S+|www\.\S+')
    return pattern.sub('', text)
```

## remove punctuations

```python
import string
#slower method
def rem_punc1(text):
    for char in text:
        if char in string.punctuation:
            text = text.replace(char, '')
    return text


import time
start_time = time.time()
df['review'].apply(rem_punc1)
time_taken1 = time.time() - start_time
time_taken1
```

```python
#faster method
def rem_punc2(text):
    return text.translate(str.maketrans('','',string.punctuation))


import time
start_time = time.time()
df['review'].apply(rem_punc2)
time_taken2 = time.time() - start_time
time_taken2
```

```python
time_taken1/time_taken2
```

```python
df['review'] = df['review'].apply(rem_punc2)
```

```python
def correct_chat_words(text):
    new_text = []
    for word in text.split():
        word = word.replace(',','')
        word = word.upper()
        if word in chat_abbreviations:
            new_text.append(chat_abbreviations[word])
        else:
            new_text.append(word)
    return " ".join(new_text).lower()
```

## Remove stop words :: a, an, the, is, am, are, to, with, for, of etc.

```python
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
```

```python
stopwords.words('english')
```

```python
def remove_stopwords(text):
    new_text = []
    for word in text.split():
        if word in stopwords.words('english'):
            new_text.append('')
        else:
            new_text.append(word)
    return " ".join(new_text)
```

## Handle Emojis : remove emoji OR replace emoji

```python
# Method 1: remove emojis
import re
def remove_emoji(text):
    emoji_pattern = re.compile("["
                           u"\U0001F600-\U0001F64F"  # emoticons
                           u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                           u"\U0001F680-\U0001F6FF"  # transport & map symbols
                           u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
                           u"\U00002702-\U000027B0"
                           u"\U000024C2-\U0001F251"
                           "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', text)
```

```python
df['review'] = df['review'].apply(remove_emoji)
```

## Stemming - root form of the word

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
def stem_words(text):
    return " ".join([ps.stem(word) for word in text.split()])
```

```python
text = 'walk walks walking walked'
stem_words(text)
```

```python
text = 'The incredibly incredible stories of the britishers is unadmirable.'
stem_words(text)
```

```python
# this shows that STEMMING is rule based, so the root words are incorrect
# to overcome this, use LEMMATIZATION (slower but correct)
# LEMMATIZATION gets the root words from Lexical Dictionary
```

## Lemmatization - correct root form of the words

```python
import nltk
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
wordnet_lemmatizer = WordNetLemmatizer()

text = 'He was running and eating at same time. He has bad habit of swimming after playing long hours in the Sun.'
punctuations = "?:!.,;"
sentence_words = nltk.word_tokenize(text)

for word in sentence_words:
    if word in punctuations:
        sentence_words.remove(word)
```

```python
sentence_words
print("{0:20}{1:20}".format("Word","Lemma"))
for word in sentence_words:
    print("{0:20}{1:20}".format(word,wordnet_lemmatizer.lemmatize(word,pos='v')))
#pos='v' - verb form of root word
```
