## Color Different quadrants 

### Method 1: Self method

```python
l, b, h = img_arr1.shape

# red removed
quad_0 = img_arr1[0:l//2, 0:b//2, 0]
# green removed
quad_1 = img_arr1[0:l//2, b//2:, 1]
# blue removed
quad_2 = img_arr1[l//2:, 0:b//2, 2]
#original colors
quad_3 = img_arr1[l//2:, b//2:, 0]
```

#### combining all the quadrants with different colors into one image

```python
rw1 = np.append(quad_0,quad_1, axis=1)
rw2 = np.append(quad_2,quad_3, axis=1)
combined = np.append(rw1,rw2, axis=0)
plt.imshow(combined)
```

```python
img = Image.open('s4.jpg')
img_arr = np.array(img)

l, b, h = img_arr.shape
img_arr[0:l//3,0:b//3, 0] = 0                        #0-red, red=0, remaining green+blue = cyan
img_arr[0:l//3,b//3:2*b//3, 1] = 0                   #1-green, green=0, remaining red+blue = magenta
img_arr[0:l//3,2*b//3:, 2] = 0                       #2-blue, blue=0, remaining red+green = yellow

img_arr[l//3:2*l//3,0:b//3, [0,1]] = 0               #red=0, green=0, remaining=blue
img_arr[l//3:2*l//3,b//3:2*b//3, [1,2]] = 0          #green=0, blue=0, remaining=red
img_arr[l//3:2*l//3,2*b//3:, [0,2]] = 0              #red=0, blue=0, remaining=green

img_arr[2*l//3:,0:b//3, [0,1,2]] = 0                 #red=0, green=0, blue=0, remaining=black
img_arr[2*l//3:,b//3:2*b//3] = 1                     #remaining=black
img_arr[2*l//3:,2*b//3:]                             #remaining=white

plt.imshow(img_arr)
```

```python
def bub_sort(array): 
    for i in range(0, len(array)): 
        for j in range(0, len(array) - i - 1): 
            if array[j] < array[j + 1]: 
                temp = array[j] 
                array[j] = array[j + 1] 
                array[j+1] = temp 
    return array
                
my_array = numpy.array([20,14,25,16,45,60,12,9])
```

```python
bub_sort(my_array)
```
